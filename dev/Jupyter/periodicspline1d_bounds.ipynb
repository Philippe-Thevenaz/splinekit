# Bounds
Let a periodic uniform polynomial spline of positive degree $n\in{\mathbb{N}}+1$ be the mapping $f:{\mathbb{R}}\rightarrow{\mathbb{R}},x\mapsto f(x)$ computed as

$$f(x)=\sum_{k\in{\mathbb{Z}}}\,c[{k\bmod K}]\,\beta^{n}(x-\delta x-k).$$
There, $K\in{\mathbb{N}}+1$ is its positive integer period and $\delta x\in{\mathbb{R}}$ is a global delay. The coefficients $c$ are called the spline coefficients and are free ingredients that allow one to shape the spline. The function $\beta^{n},$ with $n$ a superscript (not a power), is a B-spline of degree $n.$ It has the closed support ${\mathrm{supp}}\{\beta^{n}\}=[-\frac{n+1}{2},\frac{n+1}{2}]$ and is nonnegative, with $\forall x\in{\mathbb{R}}:\beta^{n}(x)\geq0.$ It is also made of $n+1$ polynomial pieces of unit length.

Our purpose here is to build a bounding spline $g$ of degree $m\in{\mathbb{N}}+1$ and delay $\delta y\in{\mathbb{R}}$ such that $\forall x\in{\mathbb{R}}:g(x)\geq f(x);$ moreover, we would like that $g$ be reasonably close to $f$ and that its determination be favorable, computationally. We provide now the main steps that lead to this goal. We first consider a local bound over a single interval; then, we merge the local bounds to obtain a global one.

## Bound Over a Unit Interval
Let $r\in{\mathbb{Z}}$ be some index and consider the interval ${\mathbb{X}}_{r}=[\frac{n-1}{2}+\delta x-r,\frac{n-1}{2}+\delta x-r+1).$ Over this unit-length interval, it turns out that the spline $f$ computed as above can also be computed from linear-algebra operations as

$$\forall x\in{\mathbb{X}}_{r}:f(x)={\mathbf{c}}_{r}^{{\mathsf{T}}}\,{\mathbf{W}}^{n}\,{\mathbf{v}}^{n}(\chi_{r}),$$
where ${\mathbf{c}}_{r}\in{\mathbb{R}}^{n+1}$ is a vector of $n+1$ coefficients selected from the $K$ spline coefficients of $f$ as ${\mathbf{c}}_{r}=\left(c[{\left(k-r\right)\bmod K}]\right)_{k=0}^{n},$ where ${\mathbf{v}}^{n}\in{\mathbb{R}}^{n+1}$ is a Vandermonde vector of degree $n$ such that $\forall\chi\in{\mathbb{R}}:{\mathbf{v}}^{n}(\chi)=(1,\left(\chi^{k}\right)_{k=1}^{n}),$ where $\chi_{r}\in[0,1)$ is the quantity $\chi_{r}=\left(x-\frac{n-1}{2}-\delta x+r\right),$ and where ${\mathbf{W}}^{n}\in{\mathbb{R}}^{\left(n+1\right)\times\left(n+1\right)}$ is some (invertible) matrix that depends on $n$ only.

### Bound of Higher Degree
Assume for the moment that the degree $m$ of the bounding spline $g$ is such that $m>n>0.$ We want to discover a vector ${\mathbf{u}}_{r}\in{\mathbb{R}}^{m+1}$ such that

$$\forall x\in{\mathbb{X}}_{r}:g(x)={\mathbf{u}}_{r}^{{\mathsf{T}}}\,{\mathbf{W}}^{m}\,{\mathbf{v}}^{m}(\chi_{r})\geq{\mathbf{c}}_{r}^{{\mathsf{T}}}\,{\mathbf{W}}^{n}\,{\mathbf{v}}^{n}(\chi_{r})=f(x).$$
For convenience, let us notate ${\mathbf{a}}=\left({\mathbf{W}}^{n}\right)^{{\mathsf{T}}}\,{\mathbf{c}}_{r}$ and let ${\mathbf{[\![}}\cdot\,{\mathbf{]\!]}}$ be the Iverson symbol. We claim that the choice ${\mathbf{u}}_{r}=\left({\mathbf{W}}^{m}\right)^{-{\mathsf{T}}}\,{\mathbf{h}}$ with

$${\mathbf{h}}=\left({\mathbf{[\![}}k\leq n+1\,{\mathbf{]\!]}}\,a_{k}\right)_{k=1}^{m+1}$$

satisfies our requirements. Indeed, some algebraic manipulations reveal that the difference $\left(g(x)-f(x)\right)$ vanishes over $x\in{\mathbb{X}}_{r}.$

### Bound of Same Degree
Assume now that $m=n>0.$ Then, it holds trivially that the choice ${\mathbf{u}}_{r}={\mathbf{c}}_{r}$ results in $g(x)\geq f(x).$

### Bound of Smaller Degree
Assume finally that $n>m>0.$ In this case, letting ${\mathbf{e}}_{m+1}$ be the $\left(m+1\right)$th canonical basis, we claim that

$${\mathbf{h}}=\left(a_{k}\right)_{k=1}^{m+1}+{\mathbf{e}}_{m+1}\,\sum_{k=m+1}^{n}\,\max(0,a_{k+1})$$
satisfies our requirements. Indeed, some algebraic manipulations lead to

$$\begin{eqnarray*}
{\mathbf{u}}_{r}^{{\mathsf{T}}}\,{\mathbf{W}}^{m}\,{\mathbf{v}}^{m}(\chi_{r})-{\mathbf{c}}_{r}^{{\mathsf{T}}}\,{\mathbf{W}}^{n}\,{\mathbf{v}}^{n}(\chi_{r})&=&\chi_{r}^{m}\,\sum_{k=m+1}^{n}\,\left\{\begin{array}{ll}-a_{k+1}\,\chi_{r}^{k-m},&a_{k+1}<0\\a_{k+1}\,\left(1-\chi_{r}^{k-m}\right),&0\leq a_{k+1}\end{array}\right.\\
&\geq&0.
\end{eqnarray*}$$



## Global Bound
We have found above a series of $K$ vectors ${\mathbf{u}}_{r}$ with $r\in[0\ldots K-1].$ From this series, we want to establish an array $\left[u[k]\right]_{k=0}^{K-1}$ of $K$ coefficients such that

$$\forall x\in{\mathbb{R}}:f(x)\leq g(x)=\sum_{k\in{\mathbb{Z}}}\,u[{k\bmod K}]\,\beta^{m}(x-\delta y-k).$$
First, we observe that $\left(x-\frac{n-1}{2}-\delta x+r\right)=\chi_{r}=\left(x-\frac{m-1}{2}-\delta y+r\right)$ is a condition that can be satisfied if we let $\delta y=\delta x+\frac{n-m}{2}.$
