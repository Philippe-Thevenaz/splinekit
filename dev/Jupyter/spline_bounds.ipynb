# Bounding Splines
We plot in black a random periodic spline of tunable period, degree $n,$ and delay. Then, we plot in brown a lower-bounding spline of some specified degree $m,$ and plot in blue the accompanying upper-bounding spline. The three splines are identical when it holds that $m=n.$ The places where ${\mathbb{X}}_{r}$ meets ${\mathbb{X}}_{r+1}$ coincide with the so-called knots of the spline and are shown as black dots. The samples of the splines are shown with blue stems crowned by a circle. The period limits are shown in red.

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 9 # Maximal spline degree
max_delay = 3.0 # Maximal absolute delay

# Persistent spline
f = sk.PeriodicSpline1D()

# Plot
def update_plot (
    period = 9,
    n = 2,
    m = 1,
    delay = 0.0
):
    global f # Spline

    # Update of the degree
    if f.degree != n:
        f = f.projected(degree = n, delay = delay)
    # Update of the period
    c = f.spline_coeff
    if f.period < period:
        c = np.append(c, np.random.standard_normal(period - f.period))
    else:
        c = c[ : period]
    # Spline being bounded
    f = sk.PeriodicSpline1D.from_spline_coeff(c, degree = n, delay = delay)
    # Lowerbound
    lb = f.lower_bound(degree = m)
    # Upperbound
    ub = f.upper_bound(degree = m)
    # Plot range
    lb_image = lb.image()
    ub_image = ub.image()
    plotrange = sk.interval.Closed((lb_image.infimum - 0.05, ub_image.supremum + 0.05))
    # Plot the splines
    (fig, ax) = plt.subplots()
    f.plot((fig, ax), plotpoints = 301, plotrange = plotrange, curve_fmt = "-k", curve_lw = 2)
    lb.plot((fig, ax), plotpoints = 301, curve_fmt = "-C5")
    ub.plot((fig, ax), plotpoints = 301, curve_fmt = "-C0")

    # Show the plot
    plt.show()

widgets.interactive(
    update_plot,
    period = (1, max_period),
    n = (0, max_degree),
    m = (0, max_degree),
    delay = (-max_delay, max_delay)
)
```
