# Plot of Periodic Splines
The purpose of this notebook is to give access to the style parameters of the plot of a `PeriodiSpline1D` object. The fields controlled by a tickmark decide whether a default value is used (ticked), or whether custom values are honored (unticked), in which case they can be adjusted. The samples, degree, and delay fields characterize the spline.

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_degree = 3 # Maximal spline degree
max_samples = 10 # Maximal support
max_plotpoints = 200 # Maximal number of plot points

# Initial random periodic cubic spline
s0 = sk.PeriodicSpline1D.from_spline_coeff(np.random.standard_normal(6), degree = 3, delay = np.random.standard_normal())

#widget controls
line_options = [
    ("'None'  ### NO LINE ###", 0),
    ("'-'  solid", 1),
    ("'--' dashed", 2),
    ("'-.' mixed", 3),
    ("':'  dotted", 4)
]
line_codes = [
    "None",
    "-",
    "--",
    "-.",
    ":"
]
color_options = [
    ("'C0' matplotlib blue", 0),
    ("'C1' matplotlib orange", 1),
    ("'C2' matplotlib green", 2),
    ("'C3' matplotlib red", 3),
    ("'C4' matplotlib purple", 4),
    ("'C5' matplotlib brown", 5),
    ("'C6' matplotlib pink", 6),
    ("'C7' matplotlib gray", 7),
    ("'C8' matplotlib olive", 8),
    ("'C9' matplotlib cyan", 9),
    ("'b'  named blue", 10),
    ("'g'  named green", 11),
    ("'r'  named red", 12),
    ("'c'  named cyan", 13),
    ("'m'  named magenta", 14),
    ("'y'  named yellow", 15),
    ("'k'  named black", 16),
    ("'w'  named white", 17)
]
color_codes = [
    "C0",
    "C1",
    "C2",
    "C3",
    "C4",
    "C5",
    "C6",
    "C7",
    "C8",
    "C9",
    "b",
    "g",
    "r",
    "c",
    "m",
    "y",
    "k",
    "w"
]
marker_options = [
    ("' ' ### NO MARKER ###", 0),
    ("'.' point", 1),
    ("',' pixel", 2),
    ("'o' circle", 3),
    ("'v' triangle down", 4),
    ("'^' triangle up", 5),
    ("'<' triangle left", 6),
    ("'>' triangle right", 7),
    ("'1' tri_down", 8),
    ("'2' tri_up", 9),
    ("'3' tri_left", 10),
    ("'4' tri_right", 11),
    ("'8' octagon", 12),
    ("'s' square", 13),
    ("'p' pentagon", 14),
    ("'P' plus (filled)", 15),
    ("'*' star", 16),
    ("'h' hexagon1", 17),
    ("'H' hexagon2", 18),
    ("'+' plus", 19),
    ("'x' cross", 20),
    ("'X' cross (filled)", 21),
    ("'D' diamond", 22),
    ("'d' thin diamond", 23),
    ("'|' vline", 24),
    ("'_' hline", 25)
]
marker_codes = [
    " ",
    ".",
    ",",
    "o",
    "v",
    "^",
    "<",
    ">",
    "1",
    "2",
    "3",
    "4",
    "8",
    "s",
    "p",
    "P",
    "*",
    "h",
    "H",
    "+",
    "x",
    "X",
    "D",
    "d",
    "|",
    "_"
]
plotdomain_widget = widgets.FloatRangeSlider(
    value = [-1.0, 7.0],
    min = -1.0 * max_samples,
    max = 2.0 * max_samples,
    step = 0.25,
    description = "domain:",
    readout_format = ".2f"
)
plotrange_widget = widgets.FloatRangeSlider(
    value = [-2.0, 2.0],
    min = -5.0,
    max = 5.0,
    step = 0.2,
    description = "range:",
    readout_format = ".1f"
)
curve_fmt_line_widget = widgets.Dropdown(
    options = line_options,
    value = 1,
    description = "curve:"
)
curve_fmt_color_widget = widgets.Dropdown(
    options = color_options,
    value = 0,
    description = "curve:"
)
curve_lw_widget = widgets.FloatSlider(
    value = 1.0,
    min = 0.1,
    max = 3.0,
    step = 0.1,
    description = "curve:",
    readout_format = ".1f"
)
curve_markerfmt_style_widget = widgets.Dropdown(
    options = marker_options,
    value = 3,
    description = "curve mrkr:"
)
curve_markerfmt_color_widget = widgets.Dropdown(
    options = color_options,
    value = 0,
    description = "curve mrkr:"
)
curvestem_linefmt_line_widget = widgets.Dropdown(
    options = line_options,
    value = 1,
    description = "curve stem:"
)
curvestem_linefmt_color_widget = widgets.Dropdown(
    options = color_options,
    value = 0,
    description = "curve stem:"
)
knot_marker_widget = widgets.Dropdown(
    options = marker_options,
    value = 3,
    description = "knot mrkr:"
)
knot_color_widget = widgets.Dropdown(
    options = color_options,
    value = 16,
    description = "knot color:"
)
periodbound_markerfmt_style_widget = widgets.Dropdown(
    options = marker_options,
    value = 3,
    description = "pbnd mrkr:"
)
periodbound_markerfmt_color_widget = widgets.Dropdown(
    options = color_options,
    value = 12,
    description = "pbnd mrkr:"
)
periodboundstem_linefmt_line_widget = widgets.Dropdown(
    options = line_options,
    value = 1,
    description = "curve stem:"
)
periodboundstem_linefmt_color_widget = widgets.Dropdown(
    options = color_options,
    value = 12,
    description = "curve stem:"
)

# Plot
def update_plot (
    default_plotdomain = True,
    plotdomain = [-1.0, 7.0],
    default_plotrange = True,
    plotrange = [-2.0, 2.0],
    default_curve_fmt = True,
    curve_fmt_line = 1,
    curve_fmt_color = 0,
    default_curve_lw = True,
    curve_lw = 1.0,
    default_curve_markerfmt = True,
    curve_markerfmt_style = 3,
    curve_markerfmt_color = 0,
    default_curvestem_linefmt = True,
    curvestem_linefmt_line = 1,
    curvestem_linefmt_color = 0,
    default_knot_marker = True,
    knot_marker = 3,
    default_knot_color = True,
    knot_color = 16,
    default_periodbound_markerfmt = True,
    periodbound_markerfmt_style = 3,
    periodbound_markerfmt_color = 12,
    default_periodboundstem_linefmt = True,
    periodboundstem_linefmt_line = 1,
    periodboundstem_linefmt_color = 12,
    plotpoints = 50,
    samples = 5,
    degree = 3,
    delay = 0.0
):
    global s0

    # Update of the spline
    f = s0.get_samples(s0.delay, support_length = s0.period)
    if s0.period < samples:
        f = np.append(f, np.random.standard_normal(samples - len(f)))
    else:
        f = f[ : samples]
    s0 = sk.PeriodicSpline1D.from_samples(f, degree = degree).delayed_by(delay)

    # plotdomain
    plotdomain_interval = sk.interval.Closed(plotdomain)
    if default_plotdomain:
        plotdomain_interval = sk.interval.RR()
    plotdomain_widget.disabled = default_plotdomain

    # plotrange
    plotrange_interval = sk.interval.Closed(plotrange)
    if default_plotrange:
        plotrange_interval = sk.interval.RR()
    plotrange_widget.disabled = default_plotrange

    # curve_fmt
    curve_fmt_line_string = line_codes[curve_fmt_line]
    if default_curve_fmt:
        curve_fmt_line_string = "-"
    curve_fmt_line_widget.disabled = default_curve_fmt
    curve_fmt_color_string = color_codes[curve_fmt_color]
    if default_curve_fmt:
        curve_fmt_color_string = "C0"
    curve_fmt_color_widget.disabled = default_curve_fmt

    # curve_lw
    curve_linewidth = curve_lw
    if default_curve_lw:
        curve_linewidth = 1.0
    curve_lw_widget.disabled = default_curve_lw

    # curve_markerfmt
    curve_markerfmt_style_string = marker_codes[curve_markerfmt_style]
    if default_curve_markerfmt:
        curve_markerfmt_style_string = "o"
    curve_markerfmt_style_widget.disabled = default_curve_markerfmt
    curve_markerfmt_color_string = ""
    if 0 != curve_markerfmt_style:
        curve_markerfmt_color_string = color_codes[curve_markerfmt_color]
    if default_curve_markerfmt:
        curve_markerfmt_color_string = "C0"
    curve_markerfmt_color_widget.disabled = default_curve_markerfmt

    # curvestem_linefmt
    curvestem_linefmt_line_string = line_codes[curvestem_linefmt_line]
    if default_curvestem_linefmt:
        curvestem_linefmt_line_string = "-"
    curvestem_linefmt_line_widget.disabled = default_curvestem_linefmt
    curvestem_linefmt_color_string = ""
    if 0 != curvestem_linefmt_line:
        curvestem_linefmt_color_string = color_codes[curvestem_linefmt_color]
    if default_curvestem_linefmt:
        curvestem_linefmt_color_string = "C0"
    curvestem_linefmt_color_widget.disabled = default_curvestem_linefmt

    # knot_marker
    knot_marker_string = marker_codes[knot_marker]
    if default_knot_marker:
        knot_marker_string = "o"
    knot_marker_widget.disabled = default_knot_marker

    # knot_color
    knot_color_string = color_codes[knot_color]
    if default_knot_color:
        knot_color_string = "k"
    knot_color_widget.disabled = default_knot_color

    # periodbound_markerfmt
    periodbound_markerfmt_style_string = marker_codes[periodbound_markerfmt_style]
    if default_periodbound_markerfmt:
        periodbound_markerfmt_style_string = "o"
    periodbound_markerfmt_style_widget.disabled = default_periodbound_markerfmt
    periodbound_markerfmt_color_string = ""
    if 0 != periodbound_markerfmt_style:
        periodbound_markerfmt_color_string = color_codes[periodbound_markerfmt_color]
    if default_periodbound_markerfmt:
        periodbound_markerfmt_color_string = "r"
    periodbound_markerfmt_color_widget.disabled = default_periodbound_markerfmt

    # periodboundstem_linefmt
    periodboundstem_linefmt_line_string = line_codes[periodboundstem_linefmt_line]
    if default_periodboundstem_linefmt:
        periodboundstem_linefmt_line_string = "-"
    periodboundstem_linefmt_line_widget.disabled = default_periodboundstem_linefmt
    periodboundstem_linefmt_color_string = ""
    if 0 != periodboundstem_linefmt_line:
        periodboundstem_linefmt_color_string = color_codes[periodboundstem_linefmt_color]
    if default_periodboundstem_linefmt:
        periodboundstem_linefmt_color_string = "r"
    periodboundstem_linefmt_color_widget.disabled = default_periodboundstem_linefmt

    # Plot of the spline
    s0.plot(
        plt.subplots(),
        plotdomain = plotdomain_interval,
        plotrange = plotrange_interval,
        plotpoints = plotpoints,
        curve_fmt = curve_fmt_line_string + curve_fmt_color_string,
        curve_lw = curve_linewidth,
        curve_markerfmt = curve_markerfmt_style_string +
            curve_markerfmt_color_string,
        curvestem_linefmt = curvestem_linefmt_line_string +
            curvestem_linefmt_color_string,
        knot_marker = knot_marker_string,
        knot_color = knot_color_string,
        periodbound_markerfmt = periodbound_markerfmt_style_string +
            periodbound_markerfmt_color_string,
        periodboundstem_linefmt = periodboundstem_linefmt_line_string +
            periodboundstem_linefmt_color_string
    )
    plt.show()

    # Function call
    print("import numpy as np # Library to handle data arrays")
    print("import matplotlib.pyplot as plt # Library to plot")
    print("import splinekit as sk # This library")
    print("")
    print("# Random data")
    print("c = np.random.standard_normal({})".format(samples))
    print("# Periodic one-dimensional spline")
    print(
        "s = sk.PeriodicSpline1D.from_spline_coeff(c, degree = {}, delay = {})"
        .format(degree, delay)
    )
    print("# Plot call that would generate an example styled like above")
    print("s.plot(")
    print("    plt.subplots(), # (fig, ax)")
    if not default_plotdomain:
        print("    plotdomain = sk.interval.Closed({}),".format(plotdomain))
    if not default_plotrange:
        print("    plotrange = sk.interval.Closed({}),".format(plotrange))
    print("    plotpoints = {},".format(plotpoints))
    if not default_curve_fmt:
        print("    curve_fmt = '{}',".format(
            curve_fmt_line_string + curve_fmt_color_string
        ))
    if not default_curve_lw:
        print("    curve_lw = {},".format(curve_linewidth))
    if not default_curve_markerfmt:
        print("    curve_markerfmt = '{}',".format(
            curve_markerfmt_style_string + curve_markerfmt_color_string
        ))
    if not default_curvestem_linefmt:
        print("    curvestem_linefmt = '{}',".format(
            curvestem_linefmt_line_string + curvestem_linefmt_color_string
        ))
    if not default_knot_marker:
        print("    knot_marker = '{}',".format(knot_marker_string))
    if not default_knot_color:
        print("    knot_color = '{}',".format(knot_color_string))
    if not default_periodbound_markerfmt:
        print("    periodbound_markerfmt = '{}',".format(
            periodbound_markerfmt_style_string + periodbound_markerfmt_color_string
        ))
    if not default_periodboundstem_linefmt:
        print("    periodboundstem_linefmt = '{}',".format(
            periodboundstem_linefmt_line_string + periodboundstem_linefmt_color_string
        ))
    print(")")
    print("plt.show()")

# interactive
widgets.interactive(
    update_plot,
    default_plotdomain = widgets.Checkbox(
        value = True,
        description = "plotdomain = sk.interval.RR()"
    ),
    plotdomain = plotdomain_widget,
    default_plotrange = widgets.Checkbox(
        value = True,
        description = "plotrange = sk.interval.RR()"
    ),
    plotrange = plotrange_widget,
    default_curve_fmt = widgets.Checkbox(
        value = True,
        description = "curve_fmt = '-C0'"
    ),
    curve_fmt_line = curve_fmt_line_widget,
    curve_fmt_color = curve_fmt_color_widget,
    default_curve_lw = widgets.Checkbox(
        value = True,
        description = "curve_lw = 1.0"
    ),
    curve_lw = curve_lw_widget,
    default_curve_markerfmt = widgets.Checkbox(
        value = True,
        description = "curve_markerfmt = 'oC0'"
    ),
    curve_markerfmt_style = curve_markerfmt_style_widget,
    curve_markerfmt_color = curve_markerfmt_color_widget,
    default_curvestem_linefmt = widgets.Checkbox(
        value = True,
        description = "curvestem_linefmt = '-C0'"
    ),
    curvestem_linefmt_line = curvestem_linefmt_line_widget,
    curvestem_linefmt_color = curvestem_linefmt_color_widget,
    default_knot_marker = widgets.Checkbox(
        value = True,
        description = "knot_marker = 'o'"
    ),
    knot_marker = knot_marker_widget,
    default_knot_color = widgets.Checkbox(
        value = True,
        description = "knot_color = 'k'"
    ),
    knot_color = knot_color_widget,
    default_periodbound_markerfmt = widgets.Checkbox(
        value = True,
        description = "periodbound_markerfmt = 'or'"
    ),
    periodbound_markerfmt_style = periodbound_markerfmt_style_widget,
    periodbound_markerfmt_color = periodbound_markerfmt_color_widget,
    default_periodboundstem_linefmt = widgets.Checkbox(
        value = True,
        description = "periodboundstem_linefmt = '-r'"
    ),
    periodboundstem_linefmt_line = periodboundstem_linefmt_line_widget,
    periodboundstem_linefmt_color = periodboundstem_linefmt_color_widget,
    plotpoints = widgets.IntSlider(
        value = 50,
        min = 2,
        max = max_plotpoints,
        step = 1,
        description = "plotpoints"
    ),
    samples = (1, max_samples),
    degree = (0, max_degree),
    delay = (-3.0, 3.0)
)
```
