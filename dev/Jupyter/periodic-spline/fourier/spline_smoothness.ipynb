# Fourier Smoothness of Periodic Splines
Build $200$ realizations of random periodic splines of tunable period, degree, and delay. Then, plot (in terms of the Fourier index) the modulus of the Fourier coefficients, averaged across the realizations of the splines. Observe experimentally that their decay (along the Fourier index) gets stronger when the degree of the spline increases, which reflects the notion that splines of higher degree are smoother.

```python
# Load the required libraries
import ipywidgets as widgets
import math
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 9 # Maximal spline degree
max_delay = 3.0 # Maximal absolute delay
max_nu = 20 # Maximal absolute index of the Fourier coefficients
realizations = 200 # Number of random realizations

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Plot update
def update_plot (
    period = 6,
    degree = 3,
    delay = 0.0
):
    # Create a list of realizations of random splines
    splines = [
        sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = degree,
            delay = delay
        )
        for _ in range(realizations + 1)
    ]

    # Compute the Fourier coefficients
    f_coeffs = np.array([
        np.array(
            [s.fourier_coeff(nu) for nu in range(-max_nu, max_nu + 1)],
            dtype = complex
        )
        for s in splines
    ])

    # Average (across the realizations) the absolute coefficients
    avg = np.reshape(np.average(np.absolute(f_coeffs), axis = -2), -1)

    # Do a stem plot
    (fig, ax) = plt.subplots()
    ax.set_ylim(bottom = -0.05 / math.sqrt(period), top = 1.0 / math.sqrt(period))
    abscissa = np.linspace(-max_nu, max_nu, num = 2 * max_nu + 1)
    (markerline, stemlines, baseline) = plt.stem(abscissa, avg)
    stemlines.set_lw(0.75)
    plt.show()

widgets.interactive(
    update_plot,
    period = (1, max_period),
    degree = (0, max_degree),
    delay = (-max_delay, max_delay)
)
```
