# Anti-Gradient of Periodic Splines
The spline $f$ whose anti-gradient is sought is shown as a <span style="color:#c20078">**fuchsia**</span> curve, with data samples at the integers represented with circles and stem lines. The sample at the origin, as well as its periodized replicates, is indicated by a red circle and stem line. The small black disks depict the knots of the spline. The anti-gradient is shown in the <span style="color:#1f77b4">**azure**</span> color, while the gradient of the anti-gradient is shown in <span style="color:#2ca02c">**green**</span>. Up to some value offset that depends on $E\{f\},$ the green and the fuchsia curves are identical.

It is instructive to realize that the knots of all three splines share their abscissa. This indicates that the delay associated to the anti-gradient is a half-integer away from the delay associated to the spline $f$ because the degree of one of these two splines is even when the degree of the other one is odd, and reciprocally.

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 5 # Maximal spline degree
max_delay = 8.0 # Maximal absolute delay

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Initial random periodic cubic splines with normal Gaussian coefficients
f0 = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)
f0 = f0.plus(rng.choice([-1.0, 1.0]) * rng.uniform(0.05, 0.15) - f0.mean())

# Plot
def update_plot (
    period = 6,
    degree = 1,
    delay = 0.0
):
    global f0

    # Update of the spline
    if f0.period != period:
        f0 = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f0.degree
        )
        # Slight departure from zero-mean
        f0 = f0.plus(rng.choice([-1.0, 1.0]) * rng.uniform(0.05, 0.15) - f0.mean())
    f0.degree = degree
    f0.delay = delay

    # Anti-gradient
    g0 = f0.anti_grad()
    # Gradient of the anti-gradient
    f = g0.gradient()

    # Dynamic range
    image = {f0.image(), g0.image(), f.image()}
    plotrange = sk.interval.Interval.enclosure(image)
    plotrange = sk.interval.Closed((
        plotrange.midpoint - 0.55 * plotrange.diameter,
        plotrange.midpoint + 0.55 * plotrange.diameter
    ))

    # Plot of the spline
    subplot = plt.subplots()
    f0.plot(subplot, plotrange = plotrange, plotpoints = 200 + 1, curve_fmt = "-m")
    g0.plot(subplot, plotrange = plotrange, plotpoints = 200 + 1, curve_fmt = "-C0")
    f.plot(subplot, plotrange = plotrange, plotpoints = 200 + 1, curve_fmt = "-C2")
    plt.show()

# Interaction
widgets.interactive(
    update_plot,
    period = (1, max_period),
    degree = (0, max_degree),
    delay = (-max_delay, max_delay)
)
```
