# Derivatives of Periodic Splines
We show in the <span style="color:#c20078">**fuchsia**</span> color the spline $f$ of degree $n\in{\mathbb{N}}$ whose derivative of order $m\in[0\ldots n]$ is sought. The data samples of $f$ at the integers are represented with circles and stem lines. The sample at the origin, as well as its periodized replicates, is indicated by a red circle and stem line. The small black disks depict the knots of the spline. The derivative $\frac{{\mathrm{d}}^{m}f(x)}{{\mathrm{d}}x^{m}}$ is shown in the <span style="color:#2ca02c">**green**</span> color.

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 5 # Maximal degree of the spline
max_delay = 2.0 # Maximal absolute delay
plotpoints = 200 + 1 # Number of rendered samples

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Initial random periodic cubic splines with normal Gaussian coefficients
f = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)

# Plot
def update_plot (
    degree = 3,
    order = 2,
    period = 6,
    delay = 0.0
):
    global f

    # Update of the spline
    if f.period != period:
        f = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f.degree
        )
    f.degree = degree
    f.delay = delay

    # Derivative of the specified order
    g = f.differentiated(order)

    # Dynamic range
    image = {f.image(), g.image()}
    plotrange = sk.interval.Interval.enclosure(image)
    plotrange = sk.interval.Closed((
        plotrange.midpoint - 0.55 * plotrange.diameter,
        plotrange.midpoint + 0.55 * plotrange.diameter
    ))

    # Plot of the spline
    subplot = plt.subplots()
    f.plot(
        subplot,
        plotrange = plotrange,
        plotpoints = 200 + 1,
        curve_fmt = "#eccdef" if order == 0 else "-m",
        curve_lw = 7.0 if order == 0 else 0.75
    )
    # Plot of the gradient
    g.plot(
        subplot,
        plotrange = plotrange,
        plotpoints = 200 + 1,
        curve_fmt = "-C2",
        curve_markerfmt = " ",
        curvestem_linefmt = "None",
        knot_marker = " ",
        periodbound_markerfmt = " ",
        periodboundstem_linefmt = "None"
    )
    plt.show()

# Interaction
degree_widget = widgets.IntSlider(min = 0, max = max_degree, value = 3)
order_widget = widgets.IntSlider(min = 0, max = 3, value = 2)
def update_order_range (
    *args
):
    order_widget.max = degree_widget.value
degree_widget.observe(update_order_range, "value")
widgets.interactive(
    update_plot,
    degree = degree_widget,
    order = order_widget,
    period = (1, max_period),
    delay = (-max_delay, max_delay)
)
```
