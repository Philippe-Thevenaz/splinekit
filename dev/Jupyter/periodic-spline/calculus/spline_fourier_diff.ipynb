# Fourier Coefficients of Periodic-Splines Derivatives
We show in the <span style="color:#c20078">**fuchsia**</span> color the spline $f$ of degree $n\in{\mathbb{N}}$ whose derivative of order $m\in[0\ldots n]$ is sought. The data samples of $f$ at the integers are represented with circles and stem lines. The sample at the origin, as well as its periodized replicates, is indicated by a red circle and stem line. The small black disks depict the knots of the spline. The derivative $\frac{{\mathrm{d}}^{m}f(x)}{{\mathrm{d}}x^{m}}$ is shown in the <span style="color:#2ca02c">**green**</span> color.

```python
# Load the required libraries
import cmath
import ipywidgets as widgets
import math
import matplotlib.lines as lines
import matplotlib.patches as patches
from matplotlib.path import Path
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 9 # Maximal degree of the spline
max_delay = 2.0 # Maximal absolute delay
max_coeff = 12 # Maximal absolute Fourier index

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Initial random periodic cubic splines with normal Gaussian coefficients
f = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)

# Plot
def update_plot (
    degree = 3,
    order = 2,
    period = 6,
    delay = 0.0
):
    global f

    # Update of the spline
    if f.period != period:
        f = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f.degree
        )
    f.degree = degree
    f.delay = delay

    # Differentiation applied in the Fourier-series domain
    def jw (
        nu
    ):
        if 0 == order:
            return(1.0)
        return(((0 + 1j) * nu * 2.0 * np.pi / period) ** order)
    ff = np.array(
        [jw(nu) * f.fourier_coeff(nu) for nu in range(-max_coeff, max_coeff + 1)],
        dtype = complex
    )

    # Differentiation applied in the native (time or space) domain
    g = f.differentiated(order)
    fg = np.array(
        [g.fourier_coeff(nu) for nu in range(-max_coeff, max_coeff + 1)],
        dtype = complex
    )

    # Modulus of the pairwise difference between all Fourier coefficients
    da = np.array(
        [[abs(ff[k1] - fg[k2]) for k2 in range(len(fg))] for k1 in range(len(ff))],
        dtype = float
    )
    a0 = np.median(da)
    s0 = np.std(da, ddof = 1.0)

    # Plot of the pairwise difference between all Fourier coefficients
    (fig, ax) = plt.subplots()
    ax.set_xlim(left = -max_coeff, right = max_coeff)
    ax.set_ylim(bottom = -max_coeff, top = max_coeff)
    ax.set_aspect(1.0)
    ax.spines[:].set_color("None")
    plt.tick_params(bottom = False, labelbottom = False, left = False, labelleft = False)
    r = 0.48
    for k1 in range(len(ff)):
        for k2 in range(len(fg)):
            if cmath.isclose(
                0.0,
                da[k1, k2],
                rel_tol = math.sqrt(math.ulp(1.0)),
                abs_tol = math.sqrt(math.ulp(1.0))
            ): # Plot identical coefficients in black
                ax.add_patch(patches.Circle(
                    (k1 - max_coeff, k2 - max_coeff),
                    r,
                    color = "k"
                ))
            else: # Plot non-identical coefficients in color
                a = min(max(-1.0, (da[k1, k2] - a0) / s0), 1.0)
                ax.add_patch(patches.Circle(
                    (k1 - max_coeff, k2 - max_coeff),
                    r,
                    color = (
                        0.0 if a < 0.0 else a, # Above median in reddish tint
                        1.0 - abs(a), # Near median in greenish tint
                        -a if a < 0.0 else 0.0, # Below median in bluish tint
                        0.5
                    )
                ))
                ph = cmath.phase(ff[k1] - fg[k2])
                ax.add_line(lines.Line2D(
                    [k1 - max_coeff, k1 - max_coeff + 0.8 * r * math.cos(ph)],
                    [k2 - max_coeff, k2 - max_coeff + 0.8 * r * math.sin(ph)],
                    lw = 0.75,
                    color = "white"
                ))
    plt.show()

# Interaction
degree_widget = widgets.IntSlider(min = 0, max = max_degree, value = 3)
order_widget = widgets.IntSlider(min = 0, max = 3, value = 2)
def update_order_range (
    *args
):
    order_widget.max = degree_widget.value
degree_widget.observe(update_order_range, "value")
widgets.interactive(
    update_plot,
    degree = degree_widget,
    order = order_widget,
    period = (1, max_period),
    delay = (-max_delay, max_delay)
)
```
