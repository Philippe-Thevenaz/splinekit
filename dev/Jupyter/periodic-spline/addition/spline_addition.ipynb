# Addition of Two Compatible Splines
We display the augend in the <span style="color:#1f77b4">**azure**</span> color; it is the realization of a random periodic polynomial spline of a specified period, degree, and delay. For the same period, degree, and delay, we display the addend in the <span style="color:#17becf">**aqua**</span> color; it is another realization of a random periodic polynomial spline. We then plot as a thick curve in the <span style="color:#8c564b">**sepia**</span> color the sum resulting from the addition of the addend to the augend.

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 9 # Maximal spline degree
max_delay = 3.0 # Maximal absolute delay

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Initial random periodic cubic splines
f1 = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)
f2 = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)

# Plot
def update_plot (
    period = 6,
    degree = 3,
    delay = 0.0
):
    global f1
    global f2

    # Update of the splines
    if f1.period != period:
        f1 = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f1.degree
        )
    f1.degree = degree
    f1.delay = delay
    if f2.period != period:
        f2 = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f2.degree
        )
    f2.degree = degree
    f2.delay = delay

    # Addition
    f = sk.PeriodicSpline1D.add(f1, f2)

    # Dynamic range
    image = {f1.image(), f2.image(), f.image()}
    plotrange = sk.interval.Interval.enclosure(image)
    plotrange = sk.interval.Closed((
        plotrange.midpoint - 0.55 * plotrange.diameter,
        plotrange.midpoint + 0.55 * plotrange.diameter
    ))

    # Plots
    subplot = plt.subplots()
    # Augend
    f1.plot(subplot, plotrange = plotrange, plotpoints = 200 + 1)
    # Addend
    f2.plot(
        subplot,
        plotrange = plotrange,
        plotpoints = 200 + 1,
        curve_fmt = "-C9",
        curve_markerfmt = "oC9",
    )
    # Sum
    f.plot(
        subplot,
        plotrange = plotrange,
        plotpoints = 200 + 1,
        curve_fmt = "-C5",
        curve_markerfmt = "oC5",
        curve_lw = 2.0
    )

    # Final display
    plt.show()

widgets.interactive(
    update_plot,
    period = (1, max_period),
    degree = (0, max_degree),
    delay = (-max_delay, max_delay)
)
```
