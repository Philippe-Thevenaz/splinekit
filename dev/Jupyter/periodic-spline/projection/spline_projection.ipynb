# Projection of a Spline
We display in thick <span style="color:#e0e0e0">**light gray**</span> the realization of a random periodic polynomial spline $f_{0}$ of a specified period, degree $n_{0}$, and delay $\delta x_{0};$ the red stemlines indicate the extent of one period, while the <span style="color:#a0a0a0">**dark gray**</span> dots indicate the knots of $f_{0}.$ We then plot in thinner <span style="color:#1f77b4">**blue**</span> its projection $f$ of degree $n$ and delay $\delta x.$

```python
# Load the required libraries
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
max_period = 15 # Maximal period
max_degree = 9 # Maximal spline degree
max_delay = 3.0 # Maximal absolute delay

# Initialize the generator of random numbers
rng = np.random.default_rng()

# Initial random periodic cubic spline
f0 = sk.PeriodicSpline1D.from_spline_coeff(rng.standard_normal(6), degree = 3)

# Plot
def update_plot (
    period = 6,
    degree_0 = 3,
    delay_0 = 0.0,
    degree = 1,
    delay = 0.0
):
    global f0

    # Update of the spline
    if f0.period != period:
        f0 = sk.PeriodicSpline1D.from_spline_coeff(
            rng.standard_normal(period),
            degree = f0.degree
        )
    f0.degree = degree_0
    f0.delay = delay_0

    # Projection
    f = f0.projected(degree = degree, delay = delay)

    # Plot of the spline undergoing a projection
    subplot = plt.subplots()
    f0.plot(
        subplot,
        plotpoints = 200 + 1,
        curve_fmt = "#e0e0e0",
        curve_lw = 7.0,
        curve_markerfmt = "oC7",
        curvestem_linefmt = "-C7",
        knot_color = "#a0a0a0"
    )

    # Plot of the projected spline
    f.plot(subplot, plotpoints = 200 + 1)

    # Final display
    plt.show()

widgets.interactive(
    update_plot,
    period = (1, max_period),
    degree_0 = (0, max_degree),
    delay_0 = (-max_delay, max_delay),
    degree = (0, max_degree),
    delay = (-max_delay, max_delay)
)
```
