# M-Scale Relation of B-Splines
A polynomial B-spline is a real function notated $\beta^{n}:{\mathbb{R}}\rightarrow{\mathbb{R}},x\mapsto\beta^{n}(x),$ where its nonnegative integer degree $n\in{\mathbb{N}}$ is typeset as a superscript (not a power). This function has many remarkable properties; here, we focus on multiresolution ones. In particular, it can be established that B-splines satisfy the M-scale equality

$$\beta^{n}({\color{blue}x})=\frac{1}{M^{n}}\,\sum_{k=0}^{\left(M-1\right)\,\left(n+1\right)}\,h^{n}_{M}[k]\,\beta^{n}({\color{blue}M\,x}+\frac{\left(M-1\right)\,\left(n+1\right)}{2}-k),$$
where $M\in{\mathbb{N}}+1$ is a positive integer that gives the scale parameter.

This equality, also called the M-scale relation, tells us that a B-spline at its nominal scale (the left-hand-side member) can be expressed as a finite weighted sum of shifted B-splines of same degree at scale $M$ (the right-hand-side member). Indeed, the free variable $\color{blue}x$ is multiplied by $\color{blue}M$ on the right, but not on the left. The weights are the coefficients $h^{n}_{M}[k]/M^{n}$ while the shifts $(\frac{\left(M-1\right)\,\left(n+1\right)}{2}-k)$ are integer-valued when at least one of $M$ or $n$ is odd.

The M-scale relation has numerous applications. It is invaluable in the development of algorithms that rely on a multiresolution approach to gradually establish the solution of certain problems. It also has fundamental links with wavelets, which allows for the decomposition of a signal into scale-related local contributions.

Here, we plot in thick green a nominal B-spline over its support, indexed by its degree. We superimpose in thin blue the shifted B-splines at scale $M$ whose sum reproduce the nominal B-spline.

```python
# Load the required libraries.
#import math
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
highest_degree = 9
highest_scale = 5
pts = 200 + 1

# Define the plot function
def m_scale_plot (
    degree = 3,
    scale = 2
):
    # Nominal support
    supp = int(sk.b_spline_support(degree).diameter)
    # Setup of the plot
    plt.xlim(-supp / 2, supp / 2)
    plt.ylim(-0.05, 1.05)
    ax = plt.gca()
    ax.spines.right.set_visible(False)
    ax.spines.top.set_visible(False)
    # Location of the samples
    abscissa = np.array([-supp / 2 + supp * k / (pts - 1) for k in range(pts)], dtype = float)
    # Nominal B-spline
    nominal_bspline_data = np.array([sk.b_spline(x, degree) for x in abscissa], dtype = float)
    plt.plot(abscissa, nominal_bspline_data, "-C2")
    # Scaled B-splines
    for k in range((scale - 1) * (degree + 1) + 1):
        scaled_bspline_data = np.array(
            [
                (scale ** (-degree)) * sk.mscale_filter(degree = degree, scale = scale)[k] *
                    sk.b_spline(scale * x + (scale - 1) * (degree + 1) / 2 - k, degree)
                for x in abscissa
            ],
            dtype = float
        )
        plt.plot(abscissa, scaled_bspline_data, "-C0", linewidth = 0.5)
    plt.show()

# Interact with the degree and the scale
interactive(m_scale_plot, degree = (0, highest_degree), scale = (1, highest_scale))
```
