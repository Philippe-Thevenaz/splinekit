# Polynomial B-Spline Derivatives
Here, we plot a B-spline $\beta^{n}$ of a given degree $n\in{\mathbb{N}}$ and all its derivatives that are finite. The plot is made over precisely the support of the spline. The colors distinguish the curves; the first few colors are <span style="color:#1f77b4">**blue**</span> for $\beta^{n},$ <span style="color:#ff7f0e">**orange**</span> for $\dot{\beta}^{n},$ <span style="color:#2ca02c">**green**</span> for $\ddot{\beta}^{n},$ and <span style="color:#d62728">**red**</span> for $\dddot{\beta}^{n}.$ Then, come <span style="color:#9467bd">**purple**</span>, <span style="color:#8c564b">**brown**</span>, <span style="color:#e377c2">**pink**</span>, <span style="color:#7f7f7f">**gray**</span>, <span style="color:#bcbd22">**olive**</span>, and <span style="color:#17becf">**cyan**</span>. After that, the colors are recycled. The higher the order of differentiation, the higher the amplitude of the differentiated spline.

```python
# Load the required libraries
from ipywidgets import interactive
import matplotlib.pyplot as plt

import splinekit as sk # This library

# Define the plot function
def b_spline_plot (
    degree = 3
):
    # Support of a B-spline
    supp = sk.b_spline_support(degree)
    # Avoid overlap and leave a margin
    # The margin serves a purpose only for the B-spline of degree 0
    period = supp.diameter + 1
    # Construct a spline from a non-overlapping periodic B-spline
    s = sk.PeriodicSpline1D.periodized_b_spline(
        period = int(period),
        degree = degree
    )
    # Get all finite derivatives
    ds = [s.differentiated(m) for m in range(degree + 1)]
    # Range of the plot
    minrange = 0
    maxrange = 0
    if 0 == degree % 4:
        minrange = ds[-1].at(1)
        maxrange = ds[-1].at(0)
    elif 1 == degree % 4:
        minrange = ds[-1].at(0.5)
        maxrange = ds[-1].at(-0.5)
    elif 2 == degree % 4:
        minrange = ds[-1].at(0)
        maxrange = ds[-1].at(1)
    elif 3 == degree % 4:
        minrange = ds[-1].at(-0.5)
        maxrange = ds[-1].at(0.5)
    # Plot the B-spline and its derivatives
    subplot = plt.subplots()
    for m in range(degree + 1):
        ds[m].plot(
            subplot,
            plotdomain = sk.interval.Closed((-0.5 * supp.diameter, 0.5 * supp.diameter)),
            plotrange = sk.interval.Closed((minrange * 1.1, maxrange * 1.1)),
            plotpoints = 200 + 1,
            curve_fmt = "-C" + str(m % 10),
            curve_markerfmt = "",
            curvestem_linefmt = "None",
            knot_marker = "",
            periodbound_markerfmt = "",
            periodboundstem_linefmt = "None"
        )

# Interact with the degree
interactive(b_spline_plot, degree = (0, 30))
```
