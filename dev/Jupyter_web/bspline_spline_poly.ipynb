# Spline Coefficients Made of Discrete Monomials
Determine the polynomial

$$\pi_{m}^{n}(x)=\sum_{k\in{\mathbb{Z}}}\,k^{m}\,\beta^{n}(x-k)$$
in the free variable $x.$ There, $n$ is the degree of the spline that one obtains by letting its spline coefficients take either a constant value for $m=0$ or the value of a monomial of degree $m\in[1\ldots n].$

```python
# Load the required libraries
from IPython.display import display
from IPython.display import Math
from ipywidgets import interactive
from sympy import latex

import splinekit as sk # This library

# Define the print function
def spline_polynomial_print (
    spline_n = 3
):
    for m in range(spline_n + 1):
        poly = sk.bsplines.spline_polynomial(
            spline_degree = spline_n,
            monomial_degree = m
        )
        display(Math(r"\sum_k\,k^{{{}}}\,\beta^{{{}}}(x-k)={}".format(m, spline_n, latex(poly))))

# Interact with the spline and monomial degrees
interactive(spline_polynomial_print, spline_n = (0, 9))
```
