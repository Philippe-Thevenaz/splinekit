# Monomial Convolved with a B-Spline
Given a nonnegative integer monomial degree $m\in{\mathbb{N}}$ and a spline degree $n\in{\mathbb{N}},$ determine the polynomial

$$\varpi_{m}^{n}(x)=\int_{\mathbb{R}}\,y^{m}\,\beta^{n}(x-y)\,{\mathrm{d}}y.$$


```python
# Load the required libraries
from IPython.display import display
from IPython.display import Math
from ipywidgets import interactive
from sympy import latex

import splinekit as sk # This library

# Define the print function
def convolve_b_spline_monomial_print (
    monomial_m = 5,
    spline_n = 3
):
    poly = sk.bsplines.convolve_b_spline_monomial(
        spline_degree = spline_n,
        monomial_degree = monomial_m
    )
    display(Math(r"\int\,y^{{{}}}\,\beta^{{{}}}(x-y)\,dy={}".format(
        monomial_m,
        spline_n,
        latex(poly))
    ))

# Interact with the spline and monomial degrees
interactive(convolve_b_spline_monomial_print, monomial_m = (0, 9), spline_n = (0, 9))
```
