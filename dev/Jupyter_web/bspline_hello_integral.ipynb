# Integral of B-Splines

Here we plot the integral of a few B-splines of low degree. The domain of integration is $\left(-\infty\right)$ on the left and the argument of the plot on the right.

```python
# Load the required libraries
import math
import matplotlib.pyplot as plt
import numpy as np

import splinekit as sk # This library

# Setup
highest_degree = 9
pts = 200 + 1
largest_supp = sk.b_spline_support(highest_degree)
supp = int(largest_supp.diameter)

# Location of the samples
abscissa = np.array(
    [-supp / 2 + supp * k / (pts - 1) for k in range(pts)],
    dtype = float
)

# One plot per degree
for degree in range(highest_degree + 1):
    # Integrated B-spline
    spline_data = np.array(
        [sk.integrated_b_spline(x, degree) for x in abscissa],
        dtype = float
    )
    plt.xlim(-supp / 2, supp / 2)
    plt.ylim(-0.05, 1.05)
    ax = plt.gca()
    ax.spines.right.set_visible(False)
    ax.spines.top.set_visible(False)
    plt.plot(
        abscissa,
        spline_data,
        "-C" + str(degree), # Distinguish degrees by color
    )
plt.show()
```
